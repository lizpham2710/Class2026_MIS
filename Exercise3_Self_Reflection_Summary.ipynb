# Exercise 3 — Self-Reflection Prompt for Improving Output

**Tools used:** ChatGPT for the original summary, self-critique, and revision; Google Colab for storing/validating the output; GitHub for notebook sharing.

**Goal:** Ask the AI to critique its own summary against specific requirements and then improve it.

### Source information used for the summary
- Segment A: average annual spending = $1,450; purchase frequency = 14; average age = 31
- Segment B: average annual spending = $450; purchase frequency = 4; average age = 38
- Segment C: average annual spending = $950; purchase frequency = 9; average age = 45

### Original summary prompt
> Summarize these three customer segments for a business manager.

### Original summary
> The customers were divided into three groups. Some customers spend more and buy more often, while other customers spend less. The business can use promotions for the better customers and advertising for the other customers.

The original summary is understandable, but it is too vague, does not use the available numbers, uses the unclear phrase “better customers,” and gives recommendations without clearly connecting them to the segment data.

## Self-reflection prompt

> Critique the original summary using these exact criteria:  
> 1. **Accuracy:** Every claim must be supported by the three segment averages provided. Do not invent facts.  
> 2. **Clarity:** Write for a nontechnical business manager and avoid vague labels such as “better customers.”  
> 3. **Focus:** Explain the practical difference between the three segments using spending and purchase frequency. Age may be mentioned only when useful.  
> 4. **Length:** The revised summary must be 90–130 words.  
> 5. **Format:** First provide 3 concise critique bullets, then provide one revised paragraph.  
> 6. **Revision:** Rewrite the summary so it fixes every problem identified in the critique.

### AI self-critique
- The original summary is too general because it does not include the actual segment averages.
- The phrase “better customers” is subjective and should be replaced with neutral, measurable descriptions.
- The recommendations are not clearly tied to each segment's spending and purchase frequency.

### Improved summary
Segment A is the highest-spending and most frequent-purchase group, averaging $1,450 in annual spending and 14 purchases. A business could focus retention or loyalty offers on this group because it already shows strong purchasing activity. Segment B has the lowest activity, averaging $450 in spending and 4 purchases, so re-engagement offers may be more appropriate than high-value rewards. Segment C is in the middle, averaging $950 in spending and 9 purchases, and may respond to offers designed to increase purchase frequency. The average ages are 31, 38, and 45 for Segments A, B, and C, but spending and purchase frequency provide the clearest differences among the groups.

```python
original_summary = """The customers were divided into three groups. Some customers spend more and buy more often, while other customers spend less. The business can use promotions for the better customers and advertising for the other customers."""

improved_summary = """Segment A is the highest-spending and most frequent-purchase group, averaging $1,450 in annual spending and 14 purchases. A business could focus retention or loyalty offers on this group because it already shows strong purchasing activity. Segment B has the lowest activity, averaging $450 in spending and 4 purchases, so re-engagement offers may be more appropriate than high-value rewards. Segment C is in the middle, averaging $950 in spending and 9 purchases, and may respond to offers designed to increase purchase frequency. The average ages are 31, 38, and 45 for Segments A, B, and C, but spending and purchase frequency provide the clearest differences among the groups."""

criteria = {
    'minimum_words': 90,
    'maximum_words': 130,
    'must_include': ['$1,450', '$450', '$950', '14', '4', '9']
}

word_count = len(improved_summary.split())
length_ok = criteria['minimum_words'] <= word_count <= criteria['maximum_words']
required_details_ok = all(item in improved_summary for item in criteria['must_include'])

print("Original summary:")
print(original_summary)
print("\nImproved summary word count:", word_count)
print("Length requirement passed:", length_ok)
print("Required numeric details included:", required_details_ok)
print("\nImproved summary:")
print(improved_summary)
```

## Improvement demonstrated
The revised version is more specific and neutral. It uses the provided averages, connects each segment to a practical business action, stays within the requested length, and avoids unsupported claims.
